```python

from braket.circuits import Circuit
from braket.devices import LocalSimulator

# Experiment 1: Bell-state quantum circuit
# Researcher: Ololade R. Odunsi

circuit = Circuit()

# Apply a Hadamard gate to qubit 0
circuit.h(0)

# Entangle qubit 0 and qubit 1
circuit.cnot(0, 1)

print("Quantum Circuit:")
print(circuit)
```

```python

# Initialize Amazon Braket's local simulator
simulator = LocalSimulator()

# Execute 1,000 measurements
task = simulator.run(circuit, shots=1000)

# Retrieve results
result = task.result()

counts = dict(result.measurement_counts)

print("Measurement Results:")
print(counts)

print("Total measurements:", sum(counts.values()))
```

```python

import json
from pathlib import Path
from datetime import datetime, timezone
from importlib.metadata import version

# Locate the project root
project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

results_dir = project_root / "results"
results_dir.mkdir(parents=True, exist_ok=True)

# Save the actual simulator results
shots = sum(counts.values())

experiment = {
    "experiment": "Bell State Quantum Circuit",
    "execution_date_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "sdk_version": version("amazon-braket-sdk"),
    "simulator": "Amazon Braket LocalSimulator",
    "qubits": 2,
    "shots": shots,
    "gates": ["Hadamard", "CNOT"],
    "measurement_counts": {
        state: int(counts.get(state, 0))
        for state in ["00", "01", "10", "11"]
    },
    "measurement_probabilities": {
        state: int(counts.get(state, 0)) / shots
        for state in ["00", "01", "10", "11"]
    },
    "notes": (
        "Introductory Bell-state experiment "
        "executed using a local quantum simulator."
    )
}

output_file = (
    results_dir / "02_bell_state_local_results.json"
)

with open(output_file, "w", encoding="utf-8") as f:
    json.dump(experiment, f, indent=4)

print("Experiment saved successfully!")
print("File location:", output_file)
```
